```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
file_name = "KTP_Raman_HeNe_1.txt"
data = np.loadtxt(file_name, skiprows=1)
probe_wavelength = 632.8  # nm
wavenumber = 1E7*(1/probe_wavelength - 1/data[:, 0]) # cm^-1
intensity = data[:, 1]
plt.plot(wavenumber, intensity)
plt.xlabel("Wavenumber (cm^-1)")
plt.ylabel("Intensity (a.u.)")
plt.title("Raman Spectrum of KTP with HeNe Laser")
plt.show()
```
